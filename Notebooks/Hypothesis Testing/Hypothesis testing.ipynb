### 1) Perceived ease of use has no impact on intention to purchase.

📌 Null Hypothesis (H₀):

There is no significant relationship between perceived ease of use and intention to purchase during a crisis.
In statistical terms:
ρ (rho) = 0

📌 Alternative Hypothesis (H₁):

There is a significant relationship between perceived ease of use and intention to purchase during a crisis.
In statistical terms:
ρ (rho) ≠ 0

```python
from scipy.stats import spearmanr
import pandas as pd

perceived_ease_of_use = (
    factors_by_section.perceived_ease_of_use.ease_of_navigation +
    factors_by_section.perceived_ease_of_use.learn_curve +
    factors_by_section.perceived_ease_of_use.clarity_of_instructions +
    factors_by_section.perceived_ease_of_use.response_time +
    factors_by_section.perceived_ease_of_use.error_handling
)

intention_column_indices = (
    factors_by_section.intention_of_purchase_online.intention_to_purchase_online
)

# Select columns by position
ease_columns = df.iloc[:,perceived_ease_of_use]
intention_columns = df.iloc[:, intention_column_indices]

# Calculate mean scores
df['ease_of_use_score'] = ease_columns.mean(axis=1)
df['purchase_intention_score'] = intention_columns.mean(axis=1)

# Hypothesis test
correlation, p_value = spearmanr(df['ease_of_use_score'], df['purchase_intention_score'])
print(f"Spearman correlation: {correlation}, p-value: {p_value}")
```

### 5) The attitude towards online shopping does not mediate the relationship between perceived ease of use and the intention to purchase online.

📌 Null Hypothesis (H₀):

Attitude towards online shopping does not mediate the relationship between perceived ease of use and intention to purchase online. In statistical terms:

The indirect effect of perceived_ease_of_use → attitude_and_perceived_risk → intention_to_purchase is not statistically significant (β_indirect = 0).
📌 Alternative Hypothesis (H₁):

Attitude towards online shopping mediates the relationship between perceived ease of use and intention to purchase online. In statistical terms:

The indirect effect of  perceived_ease_of_use → attitude_and_perceived_risk → intention_to_purchase is statistically significant (β_indirect ≠ 0).

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

# Compute scores
perceived_ease_of_use = (
    factors_by_section.perceived_ease_of_use.ease_of_navigation +
    factors_by_section.perceived_ease_of_use.learn_curve +
    factors_by_section.perceived_ease_of_use.clarity_of_instructions +
    factors_by_section.perceived_ease_of_use.response_time +
    factors_by_section.perceived_ease_of_use.error_handling
)

attitude_column_indices = (
    factors_by_section.perceived_risk.attitude +
    factors_by_section.perceived_risk.perceived_risk
)

intention_to_purchase_cols_index = (
    factors_by_section.intention_of_purchase_online.intention_to_purchase_online
)

df['ease_of_use_score'] = df.iloc[:, perceived_ease_of_use].mean(axis=1)
df['perceived_risk'] = df.iloc[:, attitude_column_indices].mean(axis=1)
df['intention_to_purchase'] = df.iloc[:, intention_to_purchase_cols_index].mean(axis=1)

# Define models
X = df['ease_of_use_score']
M = df['perceived_risk']
Y = df['intention_to_purchase']

# Add constant for intercept
X_with_const = sm.add_constant(X)
M_with_const = sm.add_constant(M)

# Model for mediator (M)
mediator_model = sm.OLS(M, X_with_const).fit()

# Model for outcome (Y) including mediator (M)
outcome_model = sm.OLS(Y, sm.add_constant(pd.concat([X, M], axis=1))).fit()

# Coefficients
# Mediator model: effect of peou on att
a = mediator_model.params[1]  # peou -> att
# Outcome model: effect of att on ipo and peou on ipo
b = outcome_model.params[2]   # att -> ipo
c_prime = outcome_model.params[1]  # peou -> ipo (direct effect)

# Indirect effect (mediation effect)
indirect_effect = a * b

# Total effect (peou on ipo)
total_effect = c_prime + indirect_effect

# Extract p-values
p_value_a = mediator_model.pvalues[1]  # p-value for peou -> att (mediator model)
p_value_b = outcome_model.pvalues[2]   # p-value for att -> ipo (outcome model)
p_value_c_prime = outcome_model.pvalues[1]  # p-value for peou -> ipo (direct effect)

# Print the results
print(f'Indirect Effect (Mediation): {indirect_effect}')
print(f'P-value for Indirect Effect (a * b): {p_value_a * p_value_b}')
print(f'Direct Effect (peou -> ipo): {c_prime}')
print(f'P-value for Direct Effect (peou -> ipo): {p_value_c_prime}')
print(f'Total Effect (peou -> ipo): {total_effect}')

# Test the mediation hypothesis (attitude as mediator)
if p_value_a < 0.05 and p_value_b < 0.05:
    print("The indirect effect is significant, suggesting that attitude mediates the relationship.")
else:
    print("The indirect effect is not significant, suggesting that attitude does not mediate the relationship.")
```

### 6) Perceived Risk mediates the relationship between Perceived Usefulness and Intention to Purchase Online during a crisis in Sri Lanka.

📌 Null Hypothesis (H₀):

Perceived Risk does not mediate the relationship between Perceived Usefulness and Intention to Purchase Online during a crisis in Sri Lanka.

In statistical terms: The indirect effect of Perceived Usefulness on Intention to Purchase Online through Perceived Risk is not significantly different from zero.

📌 Alternative Hypothesis (H₁):

Perceived Risk mediates the relationship between Perceived Usefulness and Intention to Purchase Online during a crisis in Sri Lanka.

In statistical terms: The indirect effect of Perceived Usefulness on Intention to Purchase Online through Perceived Risk is significantly different from zero.

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

# Compute scores for perceived risk, perceived usefulness, and intention to purchase
perceived_risk_index = (
    factors_by_section.perceived_risk.perceived_risk
)

perceived_usefulness_cols_index = (
    factors_by_section.perceived_usefulness.product_availability +
    factors_by_section.perceived_usefulness.convenience +
    factors_by_section.perceived_usefulness.cost_effectiveness +
    factors_by_section.perceived_usefulness.information_accessibility +
    factors_by_section.perceived_usefulness.personalization
)

intention_to_purchase_cols_index = (
    factors_by_section.intention_of_purchase_online.intention_to_purchase_online
)

df['perceived_usefulness'] = df.iloc[:, perceived_usefulness_cols_index].mean(axis=1)
df['perceived_risk'] = df.iloc[:, perceived_risk_index].mean(axis=1)
df['intention_to_purchase'] = df.iloc[:, intention_to_purchase_cols_index].mean(axis=1)

# Define models: 
X = df['ease_of_use_score']  # Independent variable (Ease of Use)
M = df['perceived_risk']     # Mediator variable (Perceived Risk)
Y = df['intention_to_purchase']  # Dependent variable (Intention to Purchase)

# Add constant for intercept to the models
X_with_const = sm.add_constant(X)
M_with_const = sm.add_constant(M)

# Model for mediator (M) – predicting mediator from independent variable
mediator_model = sm.OLS(M, X_with_const).fit()

# Model for outcome (Y) including both independent variable (X) and mediator (M)
outcome_model = sm.OLS(Y, sm.add_constant(pd.concat([X, M], axis=1))).fit()

# Coefficients:
# a = effect of X (Ease of Use) on M (Perceived Risk)
a = mediator_model.params[1]  # peou -> att
# b = effect of M (Perceived Risk) on Y (Intention to Purchase)
b = outcome_model.params[2]   # att -> ipo
# c_prime = direct effect of X (Ease of Use) on Y (Intention to Purchase)
c_prime = outcome_model.params[1]  # peou -> ipo (direct effect)

# Indirect effect (mediation effect): product of a and b
indirect_effect = a * b

# Total effect of X on Y: direct effect + indirect effect
total_effect = c_prime + indirect_effect

# Extract p-values for each path:
p_value_a = mediator_model.pvalues[1]  # p-value for peou -> att (mediator model)
p_value_b = outcome_model.pvalues[2]   # p-value for att -> ipo (outcome model)
p_value_c_prime = outcome_model.pvalues[1]  # p-value for peou -> ipo (direct effect)

# Print the results
print(f'Indirect Effect (Mediation): {indirect_effect}')
print(f'P-value for Indirect Effect (a * b): {p_value_a * p_value_b}')
print(f'Direct Effect (peou -> ipo): {c_prime}')
print(f'P-value for Direct Effect (peou -> ipo): {p_value_c_prime}')
print(f'Total Effect (peou -> ipo): {total_effect}')

# Test the mediation hypothesis (attitude as mediator)
if p_value_a < 0.05 and p_value_b < 0.05:
    print("The indirect effect is significant, suggesting that attitude mediates the relationship.")
else:
    print("The indirect effect is not significant, suggesting that attitude does not mediate the relationship.")
```
